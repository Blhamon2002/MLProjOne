```python
#Part 1 - Pull in a data set and look at it.
import pandas as pd
import numpy as np

df = pd.read_csv("world_bank_data_2025.csv")

df.info()
df.describe()

df['Log_GDP_per_Capita'] = np.log(df['GDP per Capita (Current USD)'])
numeric_df = df.select_dtypes(include=['float64', 'int64'])

corr_matrix = numeric_df.corr()

print(corr_matrix)
```

Part 2) "Did you discover interesting relations?"

There's so much to explain about how I got here. But I'll simply answer the question.

I discovered some interesting relations. Primarily when I first began looking at the data, there were really no strong correlations between GDP per Capita and any of the other features. There was a .33 correlation with public debt which I thought was interesting and the strongest of the correlations. I did use a log function to get compressed, more even spaced data when looking at GPD per Capita values as ranges are extremely varied depending on country and/or year.

Part 3) "What features would you like to be able to predict?"

I want to predict GDP per Capita (which like I said I compressed by using log, but will keep the uncompressed feature as well so that I can look at what the GDP per Capita is rather than just the compressed feature. The compressed feature of GDP per Capita allows me to see stronger correlations though.) I'd like to predict GDP per Capita by looking at Gross National Income, Tax Revenue, Government Expense, and possible Interest Rates as well.

Originally, I was going to use Public Debt (% of GPD) but there was so little data on it that it was extremely limiting to my model.

```python
#Part 4 - Clean the Data
#Admittedly I did a little cleaning in the first step because -
#I couldn't find very solid correlation without doing a log of GDP per Capita
features = [
    'Interest Rate (Real, %)',
    'Government Expense (% of GDP)',
    'Gross National Income (USD)',
    'Tax Revenue (% of GDP)',
    'GDP per Capita (Current USD)'
]

clean_df = df[features].dropna().copy()

clean_df['LogPerCapita'] = np.log(clean_df['GDP per Capita (Current USD)'])

clean_df.info()
```

```python
#Part 5 - Create Training and Test Sets
from sklearn.model_selection import train_test_split

x = clean_df[features]
y = clean_df['LogPerCapita']

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42
)

print('Training features shape:', X_train.shape)
print('Testing features shape:', X_test.shape)
```
